```python
from src.si.io.csv_file import read_csv
dataset = r"C:\Users\Álvaro Magno\Desktop\sistemas_inteligentes\sistemas_inteligentes\datasets\cpu\cpu.csv"

data = read_csv(filename=dataset, features=True, label=True)
```

```python
from src.si.model_selection.split import train_test_split
train, test = train_test_split(data, test_size=0.2, random_state=123)
```

```python
from src.si.models.ridge_regression import RidgeRegression
ridge_model = RidgeRegression(l2_penalty=1, alpha=0.001, max_iter=1000, patience=5, scale=True)
```

```python
ridge_model.fit(train)

# Fazer previsões no conjunto de teste
predictions = ridge_model.predict(test)

# Calcular o MSE (Mean Squared Error) no conjunto de teste
mse_score = ridge_model.score(test)
print(f'Mean Squared Error (MSE) on test set: {mse_score}')

# Calcular o custo final no conjunto de teste
final_cost = ridge_model.cost(test)
print(f'Final cost on test set: {final_cost}')
```
