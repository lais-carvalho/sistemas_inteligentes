```python
## Definindo o dataset com os dados do iris
from src.si.data.dataset import Dataset
dataset = Dataset(
    X=iris.data,
    y=iris.target,
    features=iris.feature_names,
    label="species"
)
```

```python
## Testando a classe criada
from src.si.decomposition.pca import PCA
selector = PCA(n_components = 2) 
selector = selector.fit(dataset)

transformed_dataset = selector.transform(dataset)

print("Features selecionadas:", transformed_dataset.features)
print(f"Variância Explicada: {selector.explained_variance}")
```

```python
## Exemplo de visuaização com gráfico
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(transformed_dataset.X[:, 0], transformed_dataset.X[:, 1], c=dataset.y, cmap='viridis')
plt.colorbar(label='Classes')
plt.title('PCA - Iris Dataset')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.show()
```
